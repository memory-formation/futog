# 00 — Data overview and sanity checks

Loads the three analysis files and reports the analytic samples. Run first.

```python
import pandas as pd, numpy as np, warnings
from scipy import stats
import statsmodels.formula.api as smf
warnings.filterwarnings('ignore')
DATA = '../data'
def load_belief(): return pd.read_excel(f'{DATA}/belief_items.xlsx')
def load_subject(): return pd.read_excel(f'{DATA}/subject_measures.xlsx')
def load_group(): return pd.read_csv(f'{DATA}/group_conversation.csv')
def rep(m,t):
    ci=m.conf_int().loc[t]
    return f'b={m.params[t]:+.3f} 95%CI[{ci[0]:+.3f}, {ci[1]:+.3f}] p={m.pvalues[t]:.4f}'
```

```python
b=load_belief(); s=load_subject(); g=load_group()
print('belief_items:', b.shape, '| subjects:', b.participant.nunique(), '| items:', b.Item.nunique())
print('  conditions:', b.condition.unique(), '| topics:', b.topic.unique())
print('subject_measures:', s.shape, '| N =', len(s))
print('  by condition:', s.condition.value_counts().to_dict())
print('group_conversation:', g.shape, '| groups with audio:', len(g))
print()
print('Per-analysis N (flags):')
print('  belief/emotion  in_belief_emotion:', int(s.in_belief_emotion.sum()))
print('  memory          in_memory:', int(s.in_memory.sum()))
print('  conversation    in_polarity_group:', int(s.in_polarity_group.sum()))
print('  connectedness (collab with conn_mean):', int(s[s.condition=="collaborative"].conn_mean.notna().sum()))
```
